                                    Pandas Series exercises

```python
import pandas as pd
print(pd.__version__)
```

                                        Series creation

Create an empty pandas Series

```python
pd.Series()
```

Given the X python list convert it to an Y pandas Series

```python
X = [1,2,3]
Y = pd.Series(X)
Y
```

Given the X pandas Series, name it 'My letters'

```python
X = pd.Series(['1','2','3'])
X.name='My letters'
X
```

Given the X pandas Series, show its values

```python
X = pd.Series([1,2,3,4,5])
X.values
```

Series indexation

Assign index names to the given X pandas Series

```python
X = pd.Series([1,2,3,4,5])
X.index = ['One','Two','Three','Four','Five']
X
```

Given the X pandas Series, show its first element

```python
X = pd.Series([1,2,3,4,5],index=['One','Two','Three','Four','Five'])
print(X[0])
print(X.iloc[0])
print(X['One'])
```

Given the X pandas Series, show its last element

```python
X = pd.Series([1,2,3,4,5],index=['O','T','Th','F','Fi'])
print(X[-1])
print(X.iloc[-1])
print(X['Fi'])
```

Given the X pandas Series, show all middle elements

```python
X = pd.Series([1,2,3,4,5],index=['O','T','Th','F','Fi'])
print(X[1:-1])
print(X.iloc[1:-1])
print(X['T':'F'])
print(X[['T','Th','F']])
```

Given the X pandas Series, show the elements in reverse position

```python
X = pd.Series([1,2,3,4],index=['A','B','C','D'])
print(X[::-1])
print(X.iloc[::-1])
```

Given the X pandas Series, show the first and last elements

```python
X = pd.Series([1,2,3,4,5],index=['A','B','C','d','e'])
print(X[[0,-1]])
print(X.iloc[[0,-1]])
print(X[['A','e']])
```

                                        Series manipulation

Convert the given integer pandas Series to float

```python
X = pd.Series([1,2,3,4])
pd.Series(X,dtype=float)
```

Reverse the given pandas Series (first element becomes last)

```python
X = pd.Series([1,2,3,4,5])
X[::-1]
```

Order (sort) the given pandas Series

```python
X = pd.Series([5,1,8,3,7],index=['O','T','Th','F','Fi'])
X.sort_values()
```

Given the X pandas Series, set the fifth element equal to 10

```python
X = pd.Series([1,2,5,7,9],index=['O','T','Th',"f",'Fi'])
X[4]=10
X
```

Given the X pandas Series, change all the middle elements to 0

```python
X = pd.Series([1,4,8,9,6,3,7],index=['O','T','Th','F','Fi','S','Se'])
X[2:5]=0
X
```

Given the X pandas Series, add 5 to every element

```python
X = pd.Series([1,2,3,4,5],index=['O','T','Th','F','Fi'])
X=X+5
X
```

                                    Series boolean arrays (also called masks)

Given the X pandas Series, make a mask showing negative elements

```python
X = pd.Series([1,2,3,4,5,-1],index=['q','w','e','r','t','y'])
X<0
```

Given the X pandas Series, get the negative elements

```python
X = pd.Series([1,-9,-5,0,2,5])
neg = X<0
X[neg]
```

Given the X pandas Series, get numbers higher than 5

```python
X = pd.Series([1,6,8,4,2,3])
five = X>5
X[five]
```

Given the X pandas Series, get numbers higher than the elements mean

```python
X = pd.Series([1,2,5,7,9])
print(X.mean())
avg = X>X.mean()
X[avg]
```

Given the X pandas Series, get numbers equal to 2 or 10

```python
X = pd.Series([1,2,2,10,1,0,10])
num = (X==2) | (X==10)
X[num]
```

                                                Logic functions

Given the X pandas Series, return True if none of its elements is zero

```python
X = pd.Series([7,8,9,3,4],index=['q','w','e','r','t'])
X.all() #Which checks whether all elements are True
```

Given the X pandas Series, return True if any of its elements is zero

```python
X = pd.Series([0,8,0,3,0],index=['q','w','e','r','t'])
X.any() #Which checks any of the element is True
```

                                        Summary statistics

Given the X pandas Series, show the sum of its elements

```python
X = pd.Series([1,4,8,9,6,3,5,7],index=['q','w','e','r','t','y','u','o'])
X.sum()
```

Given the X pandas Series, show the mean value of its elements

```python
X = pd.Series([1,5,0,9,8,7],index=['z','x','c','v','b','n'])
X.mean()
```

Given the X pandas Series, show the max value of its elements

```python
X = pd.Series([1,4,7,8,9,6,3,2],index=['a','s','d','f','g','h','j','k'])
X.max()
```
